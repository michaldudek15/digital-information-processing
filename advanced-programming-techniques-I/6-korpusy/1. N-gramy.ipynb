n-gramem nazywamy każdą sekwencję n kolejnych składowych 

sekwencje mogą się zazębiać

w przypadku analizy języka składowymi mogą być litery, sylaby lub słowa 

n-gramy pozwalają przedstawiać korpus tekstowy w postaci wektora częstości. Jest to metoda prosta i skalowalna.


**PRZYKŁADY**

**n-gramy literowe**

słowo: przetwarzanie

unigramy: p, r, z, e, t, w, a, r, z, a, n, i , e

digramy: pr, rz, ze, et, tw, wa, ar, rz, za, an, ni, ie 

trigramy: prz, rze, zet, etw, twa, war, arz, rza, zan, ani, nie 


**n-gramy sylabowe**
słowo: przetwarzanie (podział na sylaby: prze, twa, rza, nie)

digramy: przetwa, twarza, rzanie 

trigramy: przetwarza, twarzanie 

**n-gramy słowne**
Zdanie: mężny bądź, chroń pułk twój i sześć flag. 

digramy: mężny bądź, bądź chroń, chroń pułk, pułk twój, twój i, i sześć, sześć flag 

trigramy: mężny bądź chroń, bądź chroń pułk, chroń pułk twój, pułk twój i, twój i sześć, i sześć flag



# Proszę napisać program, który wygeneruje "listy" unigramów, digramów i trigramów (na literach!) dla tekstu POL/Pan Tadeusz.



```

PL = {
    'ab': 5,
    'cd': 8,
    'xy': 1
}

EN = {
    'ab': 2,
    'cd': 20,
    'xy': 8
}

```

````python
import re
from collections import Counter

def wczytaj_tekst(plik):
    with open(plik, "r") as plik:
        tekst = plik.read().lower()
        return re.sub(r"[^a-ząćęłńóśżźà ]", "", tekst)
        
def ngramy(tekst, n):
    return [tekst[i:i+n] for i in range(len(tekst)-n+1)]

def policz_ngramy(tekst, n):
    return dict(Counter(ngramy(tekst, n)))

def ranking(ngram_dict):
    return sorted(ngram_dict.items(), key=lambda x: x[1], reverse=True)


tekst = wczytaj_tekst("KORPUSY/POL/POL1 Pan Tadeusz.txt")

unigramy = policz_ngramy(tekst, 1)
digramy  = policz_ngramy(tekst, 2)
trigramy = policz_ngramy(tekst, 3)

ranking_unigramow = ranking(unigramy)
ranking_digramow = ranking(digramy)
ranking_trigramow = ranking(trigramy)

# Wyświetlenie TOP 20
print("20 najczęstszych unigramów:")
for gram, count in ranking_unigramow[:20]:
    print(gram, count)

print("\n20 najczęstszych digramów:")
for gram, count in ranking_digramow[:20]:
    print(gram, count)

print("\n20 najczęstszych trigramów:")
for gram, count in ranking_trigramow[:20]:
    print(gram, count)


print("UNIGRAMY:\n", unigramy)
print("\nDIGRAMY:\n", digramy)
print("\nTRIGRAMY:\n", trigramy)
    
````

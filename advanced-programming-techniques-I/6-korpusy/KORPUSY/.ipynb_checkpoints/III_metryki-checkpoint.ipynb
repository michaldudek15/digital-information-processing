```python
def read_samples():
    """Wczytanie testowych tekstów"""
    dir = 'KORPUSY/'
    files = 'sample.txt', 'sample2.txt'

    text = ''
    for file in files:
        if os.path.isfile(dir+file): # Sprawdź, czy to jest plik
            text += ' '+II_uczenie_modelu.preprocess_text(II_uczenie_modelu.read_file(dir+file))
    print( 'SMP', len(text) )

    ngrams = {}
    for n in (1,2,3):
        if n not in ngrams:
            ngrams[n] = {}
        ngrams[n]['SMP'] = II_uczenie_modelu.generate_ngrams(text, n)
    
    return ngrams
```

```python
def metryka_taksowkowa(ngramy1, ngramy2):
    """Oblicza metrykę taksówkową między dwoma słownikami ngramów."""
    ngramy_wszystkie = set(ngramy1.keys()) | set(ngramy2.keys())
    dystans = 0
    for ngram in ngramy_wszystkie:
        czestosc1 = ngramy1.get(ngram, 0)
        czestosc2 = ngramy2.get(ngram, 0)
        dystans += abs(czestosc1 - czestosc2)
    return dystans
```

```python
def metryka_euklidesowa(ngramy1, ngramy2):
    """Oblicza metrykę euklidesową między dwoma słownikami ngramów."""
    ngramy_wszystkie = set(ngramy1.keys()) | set(ngramy2.keys())
    dystans = 0
    for ngram in ngramy_wszystkie:
        czestosc1 = ngramy1.get(ngram, 0)
        czestosc2 = ngramy2.get(ngram, 0)
        dystans += (czestosc1 - czestosc2) ** 2
    return math.sqrt(dystans)
```

```python
def metryka_cosinusowa(ngramy1, ngramy2):
    """Oblicza metrykę cosinusową między dwoma słownikami ngramów."""
    ngramy_wszystkie = set(ngramy1.keys()) | set(ngramy2.keys())
    iloczyn_skalarny = 0
    norma1 = 0
    norma2 = 0
    for ngram in ngramy_wszystkie:
        czestosc1 = ngramy1.get(ngram, 0)
        czestosc2 = ngramy2.get(ngram, 0)
        iloczyn_skalarny += czestosc1 * czestosc2
        norma1 += czestosc1 ** 2
        norma2 += czestosc2 ** 2
    if norma1 == 0 or norma2 == 0:
        return 0  # Unikaj dzielenia przez zero
    return 1 - (iloczyn_skalarny / (math.sqrt(norma1) * math.sqrt(norma2)))
```

```python
# Przykładowe dane
ngramy = {
    "polski": {"aa": 3, "ab": 2, "ac": 1},
    "angielski": {"aa": 40, "ab": 50, "ad": 10},
    "niemiecki": {"aa": 3, "ab": 5, "ad": 1},
}
```

```python
# Przykładowe użycie
print(f"Taksówkowa (polski, angielski): {metryka_taksowkowa(ngramy['polski'], ngramy['angielski'])}")
print(f"Taksówkowa (polski, niemiecki): {metryka_taksowkowa(ngramy['polski'], ngramy['niemiecki'])}")
print(f"Taksówkowa (angielski, niemiecki): {metryka_taksowkowa(ngramy['angielski'], ngramy['niemiecki'])}")

print()

# Przykładowe użycie
print(f"Euklidesowa (polski, angielski): {metryka_euklidesowa(ngramy['polski'], ngramy['angielski'])}")
print(f"Euklidesowa (polski, niemiecki): {metryka_euklidesowa(ngramy['polski'], ngramy['niemiecki'])}")
print(f"Euklidesowa (angielski, niemiecki): {metryka_euklidesowa(ngramy['angielski'], ngramy['niemiecki'])}")
print()

# Przykładowe użycie
print(f"Cosinusowa (polski, angielski): {metryka_cosinusowa(ngramy['polski'], ngramy['angielski'])}")
print(f"Cosinusowa (polski, niemiecki): {metryka_cosinusowa(ngramy['polski'], ngramy['niemiecki'])}")
print(f"Cosinusowa (angielski, niemiecki): {metryka_cosinusowa(ngramy['angielski'], ngramy['niemiecki'])}")
```

```python
def normalizuj_ngramy(ngramy):
    """Normalizuje częstotliwości ngramów przez długość korpusu."""
    suma_czestosci = sum(ngramy.values())
    if suma_czestosci == 0:
        return ngramy  # Unikaj dzielenia przez zero
    znormalizowane_ngramy = {ngram: czestosc / suma_czestosci for ngram, czestosc in ngramy.items()}
    return znormalizowane_ngramy
```

```python
normalizowane_ngramy = {}
normalizowane_ngramy['polski'] = normalizuj_ngramy(ngramy['polski'])
normalizowane_ngramy['angielski'] = normalizuj_ngramy(ngramy['angielski'])
normalizowane_ngramy['niemiecki'] = normalizuj_ngramy(ngramy['niemiecki'])

# Przykładowe użycie
print(f"Taksówkowa (polski, angielski): {metryka_taksowkowa(normalizowane_ngramy['polski'], normalizowane_ngramy['angielski'])}")
print(f"Taksówkowa (polski, niemiecki): {metryka_taksowkowa(normalizowane_ngramy['polski'], normalizowane_ngramy['niemiecki'])}")
print(f"Taksówkowa (angielski, niemiecki): {metryka_taksowkowa(normalizowane_ngramy['angielski'], normalizowane_ngramy['niemiecki'])}")

print()

# Przykładowe użycie
print(f"Euklidesowa (polski, angielski): {metryka_euklidesowa(normalizowane_ngramy['polski'], normalizowane_ngramy['angielski'])}")
print(f"Euklidesowa (polski, niemiecki): {metryka_euklidesowa(normalizowane_ngramy['polski'], normalizowane_ngramy['niemiecki'])}")
print(f"Euklidesowa (angielski, niemiecki): {metryka_euklidesowa(normalizowane_ngramy['angielski'], normalizowane_ngramy['niemiecki'])}")
print()

# Przykładowe użycie
print(f"Cosinusowa (polski, angielski): {metryka_cosinusowa(normalizowane_ngramy['polski'], normalizowane_ngramy['angielski'])}")
print(f"Cosinusowa (polski, niemiecki): {metryka_cosinusowa(normalizowane_ngramy['polski'], normalizowane_ngramy['niemiecki'])}")
print(f"Cosinusowa (angielski, niemiecki): {metryka_cosinusowa(normalizowane_ngramy['angielski'], normalizowane_ngramy['niemiecki'])}")
```

```python
def normalizuj_ngramy_max(ngramy):
    """Normalizuje częstotliwości ngramów przez maksymalną częstotliwość."""
    max_czestosc = max(ngramy.values())
    if max_czestosc == 0:
        return ngramy  # Unikaj dzielenia przez zero
    znormalizowane_ngramy = {ngram: czestosc / max_czestosc for ngram, czestosc in ngramy.items()}
    return znormalizowane_ngramy
```

```python
normalizowane_max_ngramy = {}
normalizowane_max_ngramy['polski'] = normalizuj_ngramy_max(ngramy['polski'])
normalizowane_max_ngramy['angielski'] = normalizuj_ngramy_max(ngramy['angielski'])
normalizowane_max_ngramy['niemiecki'] = normalizuj_ngramy_max(ngramy['niemiecki'])

# Przykładowe użycie
print(f"Taksówkowa (polski, angielski): {metryka_taksowkowa(normalizowane_max_ngramy['polski'], normalizowane_max_ngramy['angielski'])}")
print(f"Taksówkowa (polski, niemiecki): {metryka_taksowkowa(normalizowane_max_ngramy['polski'], normalizowane_max_ngramy['niemiecki'])}")
print(f"Taksówkowa (angielski, niemiecki): {metryka_taksowkowa(normalizowane_max_ngramy['angielski'], normalizowane_max_ngramy['niemiecki'])}")

print()

# Przykładowe użycie
print(f"Euklidesowa (polski, angielski): {metryka_euklidesowa(normalizowane_max_ngramy['polski'], normalizowane_max_ngramy['angielski'])}")
print(f"Euklidesowa (polski, niemiecki): {metryka_euklidesowa(normalizowane_max_ngramy['polski'], normalizowane_max_ngramy['niemiecki'])}")
print(f"Euklidesowa (angielski, niemiecki): {metryka_euklidesowa(normalizowane_max_ngramy['angielski'], normalizowane_max_ngramy['niemiecki'])}")
print()

# Przykładowe użycie
print(f"Cosinusowa (polski, angielski): {metryka_cosinusowa(normalizowane_max_ngramy['polski'], normalizowane_max_ngramy['angielski'])}")
print(f"Cosinusowa (polski, niemiecki): {metryka_cosinusowa(normalizowane_max_ngramy['polski'], normalizowane_max_ngramy['niemiecki'])}")
print(f"Cosinusowa (angielski, niemiecki): {metryka_cosinusowa(normalizowane_max_ngramy['angielski'], normalizowane_max_ngramy['niemiecki'])}")
```

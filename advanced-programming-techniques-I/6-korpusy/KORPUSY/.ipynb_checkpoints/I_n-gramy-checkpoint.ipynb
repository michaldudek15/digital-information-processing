```python
import re
import regex
from pprint import pprint
```

```python
def read_file(filename):
    """Wczytuje zawartość pliku i zwraca jako tekst."""
    with open(filename, 'r', encoding='utf-8') as file:
        return file.read()
```

```python
def preprocess_text(text):
    """Czyści tekst, usuwając znaki specjalne i konwertując go do małych liter."""
    text = text.lower().strip() # zamienia wszystkie litery na małe i obcina początkowe i końcowe białe znaki
    text = regex.sub(r'[^\p{L}\s]+', ' ', text, flags=re.UNICODE) # zostawia tyłko litery i białe znaki
    text = re.sub(r'\s+', ' ', text) # zamienia wszystkie białe znaki na pojedyncze spacje
    return text
```

```python
def generate_list_of_ngrams(text, n):
    """Generuje n-gramy z tekstu."""
    ngrams = []
    for i in range(len(text)-n+1): # iterujemy po części napisu
        scrap = text[i:i+n] # przy każdym przejściu wycinamy fragment długości n-gramu
        if ' ' not in scrap[1:-1]: # nie zawiera spacji w środku napisu
            if not scrap.isspace(): # nie są to same spacje 
                ngrams.append( scrap )
    return ngrams
```

```python
def first(filename):
    text = read_file(filename)
    preprocessed_text = preprocess_text(text)
    # print(text)
    # print(preprocessed_text)
    
    unigrams = generate_list_of_ngrams(preprocessed_text, 1)
    bigrams = generate_list_of_ngrams(preprocessed_text, 2)
    trigrams = generate_list_of_ngrams(preprocessed_text, 3)
    
    print("Unigramy:", unigrams)
    print("Bigramy:", bigrams)
    print("Trigramy:", trigrams)
```

```python
first('KORPUSY/sample.txt')
```

```python
def generate_ngrams(text, n):
    """Generuje n-gramy z tekstu."""
    ngrams = {}
    for i in range(len(text)-n+1):
        scrap = text[i:i+n]
        if ' ' not in scrap[1:-1]: 
            if not scrap.isspace():
                ngrams[scrap] = ngrams.get(scrap, 0) + 1
    return ngrams
```

```python
def second(filename):
    text = read_file(filename)
    preprocessed_text = preprocess_text(text)
    
    unigrams = generate_ngrams(preprocessed_text, 1)
    bigrams = generate_ngrams(preprocessed_text, 2)
    trigrams = generate_ngrams(preprocessed_text, 3)

    print("Unigramy:")
    pprint( sorted(unigrams.items(), key=lambda item: item[1], reverse=True) )
    print("Bigramy:")
    pprint( sorted(bigrams.items(), key=lambda item: item[1], reverse=True) )
    print("Trigramy:")
    pprint( sorted(trigrams.items(), key=lambda item: item[1], reverse=True) )
```

```python
second('KORPUSY/sample.txt')
```

```python
import re
import regex
import glob
import os
import marshal
import time
```

```python
def read_file(filename):
    """Wczytuje zawartość pliku i zwraca jako tekst."""
    with open(filename, 'r', encoding='utf-8') as file:
        return file.read()
```

```python
def preprocess_text(text):
    """Czyści tekst, usuwając znaki specjalne i konwertując do małych liter."""
    text = text.lower().strip() # zamienia wszystkie litery na małe i obcina początkowe i końcowe białe znaki
    text = regex.sub(r'[^\p{L}\s]', '', text, flags=re.UNICODE) # zostawia tyłko litery i białe znaki
    text = re.sub(r'\s+', ' ', text) # zamienia wszystkie białe znaki na pojedyncze spacje
    return text
```

```python
def read_language(language):
    """Wczytuje wszystkie pliki z danego języka"""
    files = glob.glob('KORPUSY/'+language+'/*') # wszystkie elementy w podkatalogu

    text = ''
    for file in files:
        if os.path.isfile(file): # Sprawdź, czy to jest plik
            text += ' '+preprocess_text(read_file(file))

    return text
```

```python
def detect_languages():
    """Wykrywa jakie mamy dostępne języki"""
    dirs = glob.glob('KORPUSY/*') # wszystkie elementy w podkatalogu

    language_list = []
    for dir in dirs:
        if os.path.isdir(dir):
            language_list.append( os.path.basename(dir) )

    return language_list
    
```

```python
detect_languages()
```

```python
def generate_ngrams(text, n):
    """Generuje n-gramy z tekstu."""
    ngrams = {}
    for i in range(len(text)-n+1):
        scrap = text[i:i+n]
        if ' ' not in scrap[1:-1]: 
            if not scrap.isspace():
                ngrams[scrap] = ngrams.get(scrap, 0) + 1
    return ngrams
```

```python
def read_all():
    languages = detect_languages()

    ngrams = {}
    
    for language in languages:
        text = read_language(language)
        print( language, len(text) )
        for n in (1,2,3):
            if n not in ngrams:
                ngrams[n] = {}
            # print( generate_ngrams(text, 2) )
            ngrams[n][language] = generate_ngrams(text, n)
    
    return ngrams
```

```python
def save_marshal(data):
    """Zapisuje dane za pomocą marshal, jeśli plik nie istnieje."""
    file = 'KORPUSY/marshal'
    if not os.path.exists(file):
        with open(file, 'wb') as f:
            marshal.dump(data, f)
            print(f"Dane zapisano do {file}")
    else:
        print(f"Plik {file} już istnieje. Dane nie zostały zapisane.")

def read_marshal():
    """Wczytuje dane za pomocą marshal z pliku."""
    file = 'KORPUSY/marshal'
    if os.path.exists(file):
        try:
            with open(file, 'rb') as f:
                return marshal.load(f)
        except (EOFError, OSError):
            print(f"Błąd odczytu pliku {file}.")
            return None
    else:
        return None
```

```python
ngrams = read_marshal()

start_time = time.time()  # Początek pomiaru

if ngrams is None:
    # Jeśli plik nie istnieje lub odczyt się nie powiódł, zapisz dane
    ngrams = read_all()
    save_marshal( ngrams )
else:
    print("N-gramy wczytano z pliku")

end_time = time.time()  # Koniec pomiaru
print(f"Czas wykonania: {end_time - start_time:.4f} sekundy")
```
